```python
import nltk
from nltk.tokenize import (
    WhitespaceTokenizer,
    wordpunct_tokenize,
    TreebankWordTokenizer,
    TweetTokenizer,
    MWETokenizer
)
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk import pos_tag

nltk.download('punkt')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
```

```python
text = """During my summer vacation, I visited Vadodara with my friends.
We enjoyed walking on the historical palaces, eating delicious Gujarati food,
and watching the sunset on a mountain. I felt very relaxed and happy throughout the trip.
The people were friendly, making the vacation memorable."""
print("Original Text:\n", text)
```

```python
wt = WhitespaceTokenizer()
print("Whitespace Tokenization:", wt.tokenize(text))
```

```python
print("Punctuation-based Tokenization:", wordpunct_tokenize(text))
```

```python
tbt = TreebankWordTokenizer()
print("Treebank Tokenization:", tbt.tokenize(text))
```

```python
tweet_tokenizer = TweetTokenizer()
print("Tweet Tokenization:", tweet_tokenizer.tokenize(text))
```

```python
mwe = MWETokenizer([('Vadodara', 'Gujarat'), ('summer', 'vacation')], separator='_')
print("MWE Tokenization:", mwe.tokenize(wordpunct_tokenize(text)))
```

```python
tokens = wordpunct_tokenize(text)

# Porter Stemmer
porter = PorterStemmer()
print("Porter Stemmer:", [porter.stem(word) for word in tokens])

# Snowball Stemmer
snowball = SnowballStemmer("english")
print("Snowball Stemmer:", [snowball.stem(word) for word in tokens])
```

```python
lemmatizer = WordNetLemmatizer()


print("Lemmatization:", [lemmatizer.lemmatize(word) for word in tokens])
```
